```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import seaborn as sns
from scipy import stats
import warnings
import os
 
warnings.filterwarnings("ignore")
```

```python
OUTPUT_DIR = "eda_outputs"
os.makedirs(OUTPUT_DIR, exist_ok=True)
 

DATA_PATH = "China DBM WP1 Full Dataset 2.csv"
 
plt.rcParams.update({
    "figure.dpi": 150,
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.titlesize": 12,
    "axes.labelsize": 10,
    "xtick.labelsize": 9,
    "ytick.labelsize": 9,
    "font.family": "sans-serif",
})
 
PALETTE = {
    "diet":    "#EF9F27",  # amber
    "anthro":  "#D85A30",  # coral
    "cog":     "#378ADD",  # blue
    "demo":    "#888780",  # gray
    "missing": "#E24B4A",  # red
}
```

```python
#1. Load the data
print("Loading dataset...")
df = pd.read_csv(DATA_PATH, low_memory=False)
print(f"  Shape: {df.shape[0]:,} rows × {df.shape[1]} columns")
```

```python
#2. Defining variable groups
NUTRIENT_COLS = [
    "Energy", "Protein", "Fat", "CHO", "Ca", "Fe", "Zn", "Se",
    "Vit_A", "Vit_C", "Riboflavin", "Thiamin",
]



COG_COLS = {
    "ASQ communication":  "ASQcomT",
    "ASQ gross motor":    "ASQGmoveT",
    "ASQ fine motor":     "ASQFmoveT",
    "ASQ problem-solving":"ASQSproT",
    "ASQ social":         "ASQSocialT",
    "SSRT (speed)":       "SSRT",
    "PP score (memory)":  "PPscore",
    "WS score (prob-solv)":"WSscore",
    "PM total (memory)":  "PMtotal",
}

ANTHRO_COLS = ["haz", "waz", "baz", "whz", "BMI"]
MALNUT_COLS = ["stunting3", "wasting3", "thinness3"]

DEMO_COLS   = ["gender", "agey", "agegroup3", "region", "income1",
               "caretakeredu", "fatheredu", "motheredu"]

FOOD_FREQ_COLS = [c for c in df.columns if c.startswith("A") and c[1:].isdigit()]

INFANT_DIET_COLS = [
    "DQBmilk", "DQFMP", "DQmilk", "DQsoup", "DQSSB", "DQSfood",
    "DQcreal", "DQtubers", "DQdark", "DQother", "DQmeat", "DQegg",
    "DQdairy", "DQbeans", "DQnuts", "DQsupplement", "DQothersup",
    "mddbreastmilk", "mddgrains", "mddbeans", "mdddairy",
    "mddflesh", "mddeggs", "mddarichfv", "mddotherfv", "evbf",
]
```

```python
#3a.Missing data overview
print("\n[1/7] Missing data overview...")
 
miss_pct = df.isnull().mean().sort_values(ascending=False) * 100
 
# Only show columns with >5% missing
miss_high = miss_pct[miss_pct > 5].head(40)
 
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
fig.suptitle("Missing data overview", fontsize=13, fontweight="bold", y=1.01)
 
# Left: overall by variable group
groups = {
    "Nutrients": NUTRIENT_COLS,
    "Cognitive": list(COG_COLS.values()),
    "Anthro": ANTHRO_COLS + MALNUT_COLS,
    "Demographics": DEMO_COLS,
    "Food freq (A*)": FOOD_FREQ_COLS[:30],
}
group_miss = {g: df[cols].isnull().mean().mean() * 100
              for g, cols in groups.items() if any(c in df.columns for c in cols)}
colors = [PALETTE["diet"], PALETTE["cog"], PALETTE["anthro"],
          PALETTE["demo"], PALETTE["diet"]]
axes[0].barh(list(group_miss.keys()), list(group_miss.values()),
             color=colors[:len(group_miss)], edgecolor="white")
axes[0].axvline(20, color="red", linestyle="--", linewidth=0.8, alpha=0.6, label="20% threshold")
axes[0].set_xlabel("Mean missing (%)")
axes[0].set_title("Average missingness by variable group")
axes[0].legend(fontsize=8)
 
# Right: top individual variables with most missing
miss_high.head(30).plot.barh(ax=axes[1], color=PALETTE["missing"], edgecolor="white")
axes[1].set_xlabel("Missing (%)")
axes[1].set_title("Top 30 variables with highest missingness")
 
plt.tight_layout()
plt.savefig(f"{OUTPUT_DIR}/01_missing_data.png", bbox_inches="tight")
plt.show()
plt.close()
print("  Saved: 01_missing_data.png")
```

```python
#3b. Missing data overview — BY AGE GROUP (explaining the pattern above)
print("\n[1b/7] Missing data overview, split by age group...")
print("  (This shows WHY some variable groups look highly missing overall —")
print("   it's mostly because different age groups got different instruments,")
print("   not random data loss.)")

# Map agetype codes to readable labels
agetype_labels = {"1": "Infants (6-24m)", "2": "Kindergarten (2-5y)", "3": "Primary school (6-13y)"}
df["agetype_clean"] = df["agetype"].astype(str).str.strip().map(agetype_labels)

groups_for_age_check = {
    "Nutrients":       NUTRIENT_COLS,
    "Food freq (A*)":  FOOD_FREQ_COLS,
    "Cognitive":       list(COG_COLS.values()),
    "Infant diet":     INFANT_DIET_COLS,
    "Anthro":          ANTHRO_COLS + MALNUT_COLS,
}

# Build a table: rows = variable group, columns = age group, values = % missing
age_group_order = ["Infants (6-24m)", "Kindergarten (2-5y)", "Primary school (6-13y)"]
rows = []
for gname, cols in groups_for_age_check.items():
    cols_present = [c for c in cols if c in df.columns]
    if not cols_present:
        continue
    row = {"Variable group": gname}
    for age_label in age_group_order:
        sub = df[df["agetype_clean"] == age_label]
        row[age_label] = sub[cols_present].isnull().mean().mean() * 100
    rows.append(row)

miss_by_age = pd.DataFrame(rows).set_index("Variable group")
print("\n  Missingness (%) by variable group x age group:")
print(miss_by_age.round(1))

# Heatmap 
fig, ax = plt.subplots(figsize=(8, 5))
sns.heatmap(miss_by_age, annot=True, fmt=".0f", cmap="RdYlGn_r",
            vmin=0, vmax=100, linewidths=0.5, cbar_kws={"label": "% missing"}, ax=ax)
ax.set_title("Missingness by variable group and age group\n"
             "(near-100% = not collected for that age group, not random loss)",
             fontsize=11, fontweight="bold")
ax.set_xlabel("")
plt.tight_layout()
plt.savefig(f"{OUTPUT_DIR}/01b_missing_by_agegroup.png", bbox_inches="tight")
plt.show()
plt.close()
print("\n  Saved: 01b_missing_by_agegroup.png")
```

```python
#4.Section 2: Sample demographics
print("\n[2/7] Sample demographics...")

print("  Data quality notes for cleaning step:")
if (df["gender"] == "No").any():
    print(f"    - gender has {sum(df['gender']=='No')} junk value(s): 'No'")
if (df["region"].astype(str).str.strip() == "Yes").any():
    print(f"    - region has {sum(df['region'].astype(str).str.strip()=='Yes')} junk value(s): 'Yes'")
unmapped_edu = df["caretakeredu"][~df["caretakeredu"].isin([
    "Did not graduate from primary school", "Primary school graduate",
    "Junior high school graduate",
    "High school/junior college/technical school graduate",
    "College graduate", "University graduate", "Postgraduate and above",
]) & df["caretakeredu"].notna()]
if len(unmapped_edu) > 0:
    print(f"    - caretakeredu has {len(unmapped_edu)} unmapped code(s): {unmapped_edu.unique().tolist()}")

fig = plt.figure(figsize=(16, 10))
fig.suptitle("Sample demographics (n=1,887)", fontsize=13, fontweight="bold")
gs = gridspec.GridSpec(2, 3, figure=fig, hspace=0.55, wspace=0.4)

# 2a: Gender
ax1 = fig.add_subplot(gs[0, 0])
gend = df["gender"].value_counts()
bar_colors = (list(PALETTE.values()) * 3)[:len(gend)]
ax1.bar(gend.index, gend.values, color=bar_colors, edgecolor="white")
ax1.set_title("Gender distribution")
ax1.set_ylabel("Count")
for i, (idx, val) in enumerate(zip(gend.index, gend.values)):
    ax1.text(i, val + 5, str(val), ha="center", fontsize=9)

# 2b: Age in years
ax2 = fig.add_subplot(gs[0, 1])
age_num = pd.to_numeric(df["agey"], errors="coerce")
age_num.value_counts().sort_index().plot.bar(ax=ax2, color=PALETTE["demo"], edgecolor="white")
ax2.set_title("Age distribution (years)")
ax2.set_xlabel("Age (years)")
ax2.set_ylabel("Count")
ax2.tick_params(axis="x", rotation=0)

# 2c: Region
ax3 = fig.add_subplot(gs[0, 2])
region_counts = df["region"].value_counts()
ax3.pie(region_counts.values, labels=region_counts.index,
        autopct="%1.0f%%", colors=sns.color_palette("muted", len(region_counts)),
        startangle=140, textprops={"fontsize": 8})
ax3.set_title("Region")

# 2d: Caretaker education — ordered by hierarchy, not alphabetically
ax4 = fig.add_subplot(gs[1, 0])
edu_order = ["0", "Did not graduate from primary school", "Primary school graduate",
             "Junior high school graduate",
             "High school/junior college/technical school graduate",
             "College graduate", "University graduate", "Postgraduate and above", "8"]
edu_counts = df["caretakeredu"].astype(str).value_counts()
edu_order_present = [c for c in edu_order if c in edu_counts.index]
edu_counts.reindex(edu_order_present).plot.bar(ax=ax4, color=PALETTE["demo"], edgecolor="white")
ax4.set_title("Caretaker education level")
ax4.set_xlabel("")
ax4.set_ylabel("Count")
ax4.tick_params(axis="x", rotation=45)

# 2e: Household income (income1 = actual Yuan amount) — log scale to handle outliers
ax5 = fig.add_subplot(gs[1, 1])
df["income1"].plot.hist(bins=40, ax=ax5, color=PALETTE["diet"], edgecolor="white")
ax5.set_xscale("log")
ax5.set_title("Household income (Yuan/year, log scale)")
ax5.set_xlabel("Income (Yuan, log scale)")
ax5.set_ylabel("Count")

# 2f: Age group — natural age order, not alphabetical
ax6 = fig.add_subplot(gs[1, 2])
age_order = ["6-24 months", "2-5 years", "6-13 years"]
df["agegroup3"].value_counts().reindex(age_order).plot.bar(
    ax=ax6, color=PALETTE["cog"], edgecolor="white")
ax6.set_title("Age group (agegroup3)")
ax6.set_xlabel("")
ax6.set_ylabel("Count")
ax6.tick_params(axis="x", rotation=0)

plt.tight_layout()
plt.savefig(f"{OUTPUT_DIR}/02_demographics.png", bbox_inches="tight")
plt.show()
plt.close()
print("  Saved: 02_demographics.png")
```

```python
# 5.Section 3:Nutrient intake distribution
print("\n[3/7] Nutrient distributions...")

nutr_df = df[NUTRIENT_COLS].apply(pd.to_numeric, errors="coerce")

zero_var_cols = [c for c in NUTRIENT_COLS if nutr_df[c].dropna().nunique() <= 1]
if zero_var_cols:
    print(f"  WARNING: zero-variance column(s) found: {zero_var_cols} — excluding")

print("\n  Checking for near-zero columns and implausible extreme values:")
for col in NUTRIENT_COLS:
    if col in zero_var_cols:
        continue
    s = nutr_df[col].dropna()
    if len(s) == 0:
        continue
    zero_pct = (s == 0).mean() * 100
    med = s.median()
    q99 = s.quantile(0.99)
    extreme = s[s > q99 * 5] if q99 > 0 else pd.Series(dtype=float)
    if zero_pct > 30:
        print(f"    {col}: {zero_pct:.0f}% of values are exactly 0 (median={med:.4f})")
    if len(extreme) > 0:
        print(f"    {col}: {len(extreme)} value(s) >5x the 99th percentile "
              f"(99th pct={q99:.2f}, max={s.max():.2f}) — likely data entry errors, "
              f"handle in cleaning step")

nutr_valid = [c for c in NUTRIENT_COLS
              if c not in zero_var_cols and nutr_df[c].notna().sum() > 50]
print(f"\n  Plotting {len(nutr_valid)} valid nutrient columns")

n_cols = 4
n_rows = int(np.ceil(len(nutr_valid) / n_cols))
fig, axes = plt.subplots(n_rows, n_cols, figsize=(16, n_rows * 3))
fig.suptitle("Nutrient intake distributions", fontsize=13, fontweight="bold")
axes = axes.flatten()

for i, col in enumerate(nutr_valid):
    data = nutr_df[col].dropna()
    axes[i].hist(data, bins=40, color=PALETTE["diet"], edgecolor="white", alpha=0.85)
    axes[i].set_title(col)
    axes[i].set_ylabel("Count")
    skew = data.skew()
    flag = " ⚠" if skew > 10 else ""
    axes[i].text(0.97, 0.92, f"skew={skew:.1f}{flag}", transform=axes[i].transAxes,
                 ha="right", fontsize=7, color="red" if skew > 10 else "gray")

for j in range(i + 1, len(axes)):
    axes[j].set_visible(False)

plt.tight_layout()
plt.savefig(f"{OUTPUT_DIR}/03_nutrient_distributions.png", bbox_inches="tight")
plt.show()
plt.close()
print("  Saved: 03_nutrient_distributions.png")
```

```python
# 6.Section 4: Cognitive outcomes
print("\n[4/7] Cognitive outcome distributions...")

cog_df = df[[v for v in COG_COLS.values() if v in df.columns]].apply(pd.to_numeric, errors="coerce")

print("  Checking for implausible extreme values:")
for label, col in COG_COLS.items():
    if col not in cog_df.columns:
        continue
    s = cog_df[col].dropna()
    if len(s) == 0:
        continue
    q99 = s.quantile(0.99)
    extreme = s[s > q99 * 5] if q99 > 0 else pd.Series(dtype=float)
    if len(extreme) > 0:
        print(f"    {label} ({col}): {len(extreme)} value(s) >5x the 99th percentile "
              f"(99th pct={q99:.1f}, max={s.max():.1f}) — check against valid test score range, "
              f"handle in cleaning step")

fig, axes = plt.subplots(3, 3, figsize=(15, 11))
fig.suptitle("Cognitive outcome distributions", fontsize=13, fontweight="bold")
axes = axes.flatten()

for i, (label, col) in enumerate(COG_COLS.items()):
    if col not in cog_df.columns:
        axes[i].set_visible(False)
        continue
    data = cog_df[col].dropna()
    axes[i].hist(data, bins=35, color=PALETTE["cog"], edgecolor="white", alpha=0.85)
    axes[i].set_title(label)
    axes[i].set_ylabel("Count")
    axes[i].text(0.97, 0.92, f"n={len(data):,}", transform=axes[i].transAxes,
                 ha="right", fontsize=8, color="gray")

for j in range(i + 1, len(axes)):
    axes[j].set_visible(False)

plt.tight_layout()
plt.savefig(f"{OUTPUT_DIR}/04_cognitive_outcomes.png", bbox_inches="tight")
plt.show()
plt.close()
print("  Saved: 04_cognitive_outcomes.png")
```

```python
# Section 5: Malnutrition prevalence
print("\n[5/7] Malnutrition prevalence...")

fig, axes = plt.subplots(1, 3, figsize=(16, 6))
fig.suptitle("Malnutrition prevalence by gender", fontsize=13, fontweight="bold")

# map text labels to numeric codes 
malnut_maps = {
    "stunting3": {
        "Non-stunted":               "Normal",
        "Moderate stunted (-3z -2z)":"Moderate",
        "Severely Stunted (<-3z)":   "Severe"
    },
    "wasting3": {
        "Normal":             "Normal",
        "Wasting (<-2z)":     "Wasting",
        "Overweight (>2z)":   "Overweight"
    },
    "thinness3": {
        "Normal":             "Normal",
        "Thinness (<-2z)":    "Thinness",
        "Overweight (>1z)":   "Overweight"
    }
}

malnut_titles = {
    "stunting3":  "Stunting",
    "wasting3":   "Wasting",
    "thinness3":  "Thinness",
}

# wasting3 and thinness3 are age-scoped WHO indicators, not missing at random
print("  Coverage note: wasting3 applies to younger children (weight-for-height),")
print("  thinness3 applies to older children (BMI-for-age) — these are different")
print("  WHO indicators for different age ranges, not random missing data.")

# Clean gender column
df["gender_clean"] = df["gender"].astype(str).str.strip()
df_gender = df[df["gender_clean"].isin(["Male", "Female"])].copy()

for i, (col, title) in enumerate(malnut_titles.items()):
    if col not in df.columns:
        axes[i].set_visible(False)
        continue

    # Map text → readable short label
    df_gender[col + "_mapped"] = df_gender[col].map(malnut_maps[col])
    df_plot = df_gender[["gender_clean", col + "_mapped"]].dropna()

    if df_plot.empty:
        axes[i].set_title(f"{title} (no data)")
        continue

    # Crosstab: rows = gender, cols = status label
    ct = pd.crosstab(
        df_plot["gender_clean"],
        df_plot[col + "_mapped"],
        normalize="index"
    ) * 100

    ct.T.plot.bar(
        ax=axes[i],
        color=[PALETTE["cog"], PALETTE["anthro"]],
        edgecolor="white",
        width=0.6
    )
    axes[i].set_title(f"{title}  (n={len(df_plot):,})", fontsize=12, fontweight="bold")  # FIX: show n on chart
    axes[i].set_xlabel("Status")
    axes[i].set_ylabel("% within gender")
    axes[i].legend(title="Gender", fontsize=8)
    axes[i].tick_params(axis="x", rotation=15)

plt.tight_layout()
plt.savefig(f"{OUTPUT_DIR}/05_malnutrition_prevalence.png", bbox_inches="tight")
plt.show()
plt.close()
print("  Saved: 05_malnutrition_prevalence.png")
```

```python
#SECTION 6: Nutrient correlation heatmap
print("\n[6/7] Nutrient correlation heatmap...")

corr_cols = nutr_valid + [v for v in COG_COLS.values() if v in df.columns]
corr_df = df[corr_cols].apply(pd.to_numeric, errors="coerce")

corr_df_clean = corr_df.copy()
for col in corr_df_clean.columns:
    q99 = corr_df_clean[col].quantile(0.99)
    if pd.notna(q99) and q99 > 0:
        cap = q99 * 5
        n_capped = (corr_df_clean[col] > cap).sum()
        if n_capped > 0:
            corr_df_clean.loc[corr_df_clean[col] > cap, col] = np.nan
            print(f"  Excluded {n_capped} extreme outlier(s) from {col} before correlating")

corr_matrix = corr_df_clean.corr()


n_matrix = corr_df_clean.notna().astype(int).T.dot(corr_df_clean.notna().astype(int))
low_n_mask = n_matrix < 30  # flag cells backed by fewer than 30 children

fig, ax = plt.subplots(figsize=(16, 13))
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
sns.heatmap(
    corr_matrix, mask=mask, annot=True, fmt=".2f", linewidths=0.3,
    cmap="RdBu_r", center=0, vmin=-1, vmax=1,
    annot_kws={"size": 6}, ax=ax,
    cbar_kws={"shrink": 0.6},
)

# Overlay hatching on low-n / unreliable cells (e.g. ASQ x school-age pairs)
for i in range(len(corr_matrix)):
    for j in range(len(corr_matrix)):
        if not mask[i, j] and low_n_mask.iloc[i, j]:
            ax.add_patch(plt.Rectangle((j, i), 1, 1, fill=False, hatch="///",
                                        edgecolor="black", lw=0))

ax.set_title("Pearson correlation: nutrients × cognitive outcomes\n"
             "(outliers excluded; hatched cells = fewer than 30 children overlap — "
             "e.g. ASQ vs. school-age tests, different age groups, unreliable)",
             fontsize=12, pad=14)
plt.tight_layout()
plt.savefig(f"{OUTPUT_DIR}/06_correlation_heatmap.png", bbox_inches="tight")
plt.show()
plt.close()
print("  Saved: 06_correlation_heatmap.png")
```

```python
#Section 7: Summary statistics table
print("\n[7/7] Summary statistics table...")

all_key_cols = NUTRIENT_COLS + list(COG_COLS.values()) + ANTHRO_COLS
key_df = df[[c for c in all_key_cols if c in df.columns]].apply(pd.to_numeric, errors="coerce")

key_df_clean = key_df.copy()
outlier_log = []
for col in key_df_clean.columns:
    q99 = key_df_clean[col].quantile(0.99)
    if pd.notna(q99) and q99 > 0:
        cap = q99 * 5
        n_capped = (key_df_clean[col] > cap).sum()
        if n_capped > 0:
            key_df_clean.loc[key_df_clean[col] > cap, col] = np.nan
            outlier_log.append((col, n_capped))

if outlier_log:
    print("  Excluded extreme outliers before computing summary stats:")
    for col, n in outlier_log:
        print(f"    {col}: {n} value(s) excluded")

if "PPscore" in key_df_clean.columns:
    s = key_df_clean["PPscore"]
    q99 = s.quantile(0.99)
    n_extra = (s > q99).sum()
    key_df_clean.loc[s > q99, "PPscore"] = np.nan
    print(f"  PPscore: additional {n_extra} value(s) excluded with tighter cap (>99th percentile)")

print("\n  NOTE: Missing_% reflects age-scoped survey design in several rows —")
print("  nutrients/food-frequency vars were only collected for 3-13y children,")
print("  ASQ only for infants/preschool, and SSRT/PP/WS/PM only for school-age")
print("  children. High missingness for these rows is NOT random data loss.")

summary = key_df_clean.describe(percentiles=[0.25, 0.5, 0.75]).T
summary["missing_n"]   = key_df.isnull().sum()          # FIX: missingness counted on ORIGINAL data, not outlier-excluded
summary["missing_pct"] = (key_df.isnull().mean() * 100).round(1)
summary["outliers_excluded"] = [dict(outlier_log).get(c, 0) for c in summary.index]  # NEW: transparency column
summary["skewness"]    = key_df_clean.skew().round(2)
summary = summary[["count", "mean", "std", "min", "25%", "50%", "75%", "max",
                    "missing_n", "missing_pct", "outliers_excluded", "skewness"]]
summary.columns = ["N", "Mean", "SD", "Min", "Q1", "Median", "Q3", "Max",
                   "Missing_n", "Missing_%", "Outliers_excluded", "Skewness"]
summary.to_csv(f"{OUTPUT_DIR}/07_summary_statistics.csv")
print("  Saved: 07_summary_statistics.csv")

# print of key stats
print("\n  Key cognitive outcomes (non-missing rows, outliers excluded):")
for label, col in COG_COLS.items():
    if col in key_df_clean.columns:
        d = key_df_clean[col].dropna()
        print(f"    {label:<30} n={len(d):>5}  mean={d.mean():.1f}  sd={d.std():.1f}")
```

```python
##Preprocessing
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import warnings
import os

warnings.filterwarnings("ignore")

DATA_PATH  = "China DBM WP1 Full Dataset 2.csv"
OUTPUT_DIR = "preprocessing_outputs"
os.makedirs(OUTPUT_DIR, exist_ok=True)

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)


NUTRIENT_COLS = [
    "Energy", "Protein", "Fat", "CHO", "Ca", "Fe", "Zn", "Se",
    "Vit_A", "Vit_C", "Riboflavin", "Thiamin",
]

LOG_TRANSFORM_COLS = [
    "Protein", "Fat", "Ca", "Fe", "Zn", "Se",
    "Vit_A", "Vit_C", "Riboflavin", "Thiamin",
]

# Food frequency columns
FOOD_FREQ_COLS = None  # populated after loading


INFANT_DIET_COLS = [
    "DQBmilk", "DQFMP", "DQmilk", "DQsoup", "DQSSB", "DQSfood",
    "DQcreal", "DQtubers", "DQdark", "DQother", "DQmeat", "DQegg",
    "DQdairy", "DQbeans", "DQnuts", "DQsupplement", "DQothersup",
]
INFANT_DIET_COLS = [c for c in INFANT_DIET_COLS if c in df.columns]
print(f"  Infant diet columns confirmed present: {len(INFANT_DIET_COLS)}")

print("  Recoding infant diet items to binary (1=Yes, 0=No)...")
for col in INFANT_DIET_COLS:
    df[col] = df[col].map({"Yes": 1, "No": 0})   # "Don't know", "0", "3", "4" (stray codes) -> NaN
n_recoded_infant = df[INFANT_DIET_COLS].notna().sum().sum()
print(f"    Recoded {len(INFANT_DIET_COLS)} columns, {n_recoded_infant:,} total valid responses")

COG_ASQ   = ["ASQcomT", "ASQGmoveT", "ASQFmoveT", "ASQSproT", "ASQSocialT"]
COG_OLDER = ["SSRT", "PPscore", "WSscore", "PMtotal"]
ALL_COG   = COG_ASQ + COG_OLDER


ANTHRO_COLS = ["haz", "waz", "baz", "whz", "BMI"]
MALNUT_COLS = ["stunting3", "wasting3", "thinness3"]


COVARIATE_COLS = ["gender", "agey", "agegroup3", "region",
                  "income1", "caretakeredu", "fatheredu", "motheredu"]

# Malnutrition text - ordinal integer mapping
MALNUT_MAPS = {
    "stunting3": {
        "Non-stunted": 0,
        "Moderate stunted (-3z -2z)": 1,
        "Severely Stunted (<-3z)": 2,
    },
    "wasting3": {
        "Normal": 0,
        "Wasting (<-2z)": 1,
        "Overweight (>2z)": 2,
    },
    "thinness3": {
        "Normal": 0,
        "Thinness (<-2z)": 1,
        "Overweight (>1z)": 2,
    },
}
```

```python
# Step 1: Load and initial cleaning
print("=" * 60)
print("STEP 1 — Loading and initial cleaning")
print("=" * 60)
df = pd.read_csv(DATA_PATH, low_memory=False)
print(f"  Loaded: {df.shape[0]:,} rows × {df.shape[1]} columns")

# Identify food-frequency columns
FOOD_FREQ_COLS = [c for c in df.columns if c.startswith("A") and c[1:].isdigit()]
print(f"  Food-frequency columns found: {len(FOOD_FREQ_COLS)}")

# 1a: Remove genuinely corrupted records
n_before = len(df)
df["ID_numeric_check"] = pd.to_numeric(df["ID"], errors="coerce")
corrupted_id = df["ID_numeric_check"].isna() & df["ID"].notna()
print(f"  Found {corrupted_id.sum()} row(s) with non-numeric ID (shifted/corrupted record):")
if corrupted_id.sum() > 0:
    print(f"    {df.loc[corrupted_id, 'ID'].tolist()}")
missing_id = df["ID"].isna()
print(f"  Found {missing_id.sum()} row(s) with missing ID (cannot be tracked/merged)")
df = df[~df["gender"].isin(["No", "Yes"])].copy()
df = df[~df["region"].astype(str).str.strip().isin(["Yes", "No"])].copy()
df = df[~corrupted_id.reindex(df.index, fill_value=False)].copy()  # drop shifted/non-numeric ID rows
df = df[~missing_id.reindex(df.index, fill_value=False)].copy()   # drop missing-ID rows
df = df.drop(columns=["ID_numeric_check"])
print(f"  Removed {n_before - len(df)} genuinely corrupted record(s) → {len(df):,} remaining")

#  Repair missing agegroup3 using the original agetype code, before any age-based filtering happens anywhere downstream 
agetype_to_agegroup = {"1": "6-24 months", "2": "2-5 years", "3": "6-13 years"}
missing_agegroup = df["agegroup3"].isna() & df["agetype"].notna()
print(f"  Found {missing_agegroup.sum()} row(s) with missing agegroup3 but valid agetype — repairing")
df.loc[missing_agegroup, "agegroup3"] = df.loc[missing_agegroup, "agetype"].map(agetype_to_agegroup)
still_missing = df["agegroup3"].isna().sum()
if still_missing > 0:
    print(f"  WARNING: {still_missing} row(s) still have no valid agegroup3 or agetype — "
          f"will be excluded from age-scoped analyses")

FOOD_FREQ_COLS = []
for c in df.columns:
    if c.startswith("A") and c[1:].isdigit():
        n = int(c[1:])
        if 1 <= n <= 60:
            FOOD_FREQ_COLS.append(c)
print(f"  Food-frequency columns (corrected): {len(FOOD_FREQ_COLS)} "
      f"(excluded {len([c for c in df.columns if c.startswith('A') and c[1:].isdigit()]) - len(FOOD_FREQ_COLS)} "
      f"free-text columns: top-5 vegetable/fruit lists)")

# Recode Yes/No text -> 1/0 binary 
print("  Recoding food-frequency items to binary (1=Yes, 0=No)...")
for col in FOOD_FREQ_COLS:
    df[col] = df[col].map({"Yes": 1, "No": 0})
n_recoded = df[FOOD_FREQ_COLS].notna().sum().sum()
print(f"    Recoded {len(FOOD_FREQ_COLS)} columns, {n_recoded:,} total valid responses")

# 1b: Recode malnutrition text - ordinal integers
print("  Recoding malnutrition columns to ordinal integers...")
for col, mapping in MALNUT_MAPS.items():
    if col in df.columns:
        df[col + "_num"] = df[col].map(mapping)
        matched = df[col + "_num"].notna().sum()
        print(f"    {col}: {matched:,} values matched")

# 1c: Force numeric types
for col in NUTRIENT_COLS + ANTHRO_COLS + ALL_COG + FOOD_FREQ_COLS:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

# 1d: Exclude known extreme outliers
print("  Excluding extreme outliers (before imputation):")
for col in NUTRIENT_COLS + ALL_COG:
    if col not in df.columns:
        continue
    s = df[col]
    q99 = s.quantile(0.99)
    if pd.notna(q99) and q99 > 0:
        cap = q99 * 5
        n_capped = (s > cap).sum()
        if n_capped > 0:
            df.loc[s > cap, col] = np.nan
            print(f"    {col}: {n_capped} value(s) excluded")

# Extra tight cap for PPscore specifically
if "PPscore" in df.columns:
    q99 = df["PPscore"].quantile(0.99)
    n_extra = (df["PPscore"] > q99).sum()
    if n_extra > 0:
        df.loc[df["PPscore"] > q99, "PPscore"] = np.nan
        print(f"    PPscore: additional {n_extra} value(s) excluded with tighter cap")

print(f"  Clean base shape: {df.shape}")
```

```python
# Step 2: AGE-GROUP subsetting
print("\n" + "=" * 60)
print("STEP 2 — Age-group subsetting")
print("=" * 60)


df["agegroup3"] = df["agegroup3"].astype(str).str.strip()

# ASQ subgroup: youngest children
df_asq = df[df["agegroup3"] == "6-24 months"].copy()
print(f"  ASQ subgroup (6-24 months): {len(df_asq):,} children")
print(f"    ASQ data available: {df_asq['ASQcomT'].notna().sum():,}")

# Older subgroup: kindergarten + primary school
df_older = df[df["agegroup3"].isin(["2-5 years", "6-13 years"])].copy()
print(f"  Older subgroup (2-5 + 6-13 years): {len(df_older):,} children")
print(f"    SSRT data available: {df_older['SSRT'].notna().sum():,}")
print(f"    WS score available:  {df_older['WSscore'].notna().sum():,}")
```

```python
# STEP 3: MICE IMPUTATION FOR NUTRIENTS + FOOD-FREQUENCY
print("\n" + "=" * 60)
print("STEP 3 — MICE imputation for nutrient and food-frequency variables")
print("=" * 60)

from sklearn.experimental import enable_iterative_imputer   # noqa
from sklearn.impute import IterativeImputer
from sklearn.ensemble import RandomForestRegressor

is_infant = df["agegroup3"] == "6-24 months"
print(f"  Excluding {is_infant.sum():,} infants from imputation "
      f"(nutrient/food-frequency data not collected for this age group)")


IMPUTE_COLS = [c for c in NUTRIENT_COLS + ANTHRO_COLS + FOOD_FREQ_COLS
               if c in df.columns and c != "whz"]

aux_cols = [c for c in ["agey"] if c in df.columns]
model_cols = IMPUTE_COLS + aux_cols

df_impute = df.loc[~is_infant, model_cols].copy()
print(f"  Imputing on {len(df_impute):,} children (2-13y only), "
      f"{len(IMPUTE_COLS)} columns")

print(f"\n  Running MICE (this may take longer now — 60 more columns added)...")
imputer = IterativeImputer(
    estimator=RandomForestRegressor(
        n_estimators=100, random_state=RANDOM_STATE, n_jobs=-1
    ),
    max_iter=10,
    random_state=RANDOM_STATE,
    verbose=0,
)
imputed_array = imputer.fit_transform(df_impute)
df_imputed_subset = pd.DataFrame(imputed_array, columns=model_cols, index=df_impute.index)


for col in FOOD_FREQ_COLS:
    if col in df_imputed_subset.columns:
        df_imputed_subset[col] = df_imputed_subset[col].round().clip(0, 1)

print(f"  Missingness AFTER imputation:")
remaining_miss = df_imputed_subset[IMPUTE_COLS].isnull().sum().sum()
print(f"    Total missing values (2-13y subset): {remaining_miss} (should be 0)")

df_clean = df.copy()
for col in IMPUTE_COLS:
    df_clean.loc[~is_infant, col] = df_imputed_subset.loc[~is_infant, col]

print(f"  Infants retain NaN for nutrient/food-frequency columns "
      f"(n={is_infant.sum():,}) — will use INFANT_DIET_COLS instead")
print(f"  'whz' retained as NaN for all rows — not imputed, use 'baz' instead")

for col in ["Fe", "Zn", "Ca", "Vit_A"]:
    if col in df.columns:
        df_clean[f"{col}_was_missing"] = df[col].isnull().astype(int)

print(f"  Imputation complete. Clean shape: {df_clean.shape}")
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
 
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
 
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score
import umap
```

```python
# CONFIG
INPUT_PATH = "preprocessing_outputs/X_scaled_features.csv"   
CLEAN_PATH = "preprocessing_outputs/df_clean.csv"
OUTPUT_DIR = "vae_outputs"
os.makedirs(OUTPUT_DIR, exist_ok=True)

LATENT_DIM   = 10        
HIDDEN_DIM   = 64       
BETA         = 1.0      
EPOCHS       = 500
BATCH_SIZE   = 64
LEARNING_RATE = 1e-3
RANDOM_SEED  = 42

torch.manual_seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)

sns.set_style("whitegrid")
PALETTE = sns.color_palette("Set2", 8)

print("=" * 70)
print("VAE + UMAP + GMM CLUSTERING PIPELINE")
print("=" * 70)
```

```python
# STEP 1 — LOAD DATA
print("\n[1/8] Loading scaled nutrient matrix...")

X_df = pd.read_csv(INPUT_PATH)

child_ids = X_df["ID"].copy()
feature_names = [c for c in X_df.columns if c != "ID"]
X = X_df[feature_names].values.astype(np.float32)

print(f"  Loaded: {X.shape[0]:,} children x {X.shape[1]} diet features")
print(f"  Features: {feature_names}")
print(f"  Missing values: {np.isnan(X).sum()} (should be 0)")

n_features = X.shape[1]
```

```python
# STEP 2 — DEFINE THE VAE
print("\n[2/8] Building VAE architecture...")

class VAE(nn.Module):
    def __init__(self, input_dim, hidden_dim, latent_dim):
        super().__init__()
        # Encoder
        self.enc1 = nn.Linear(input_dim, hidden_dim)
        self.enc2 = nn.Linear(hidden_dim, hidden_dim)
        self.mu_layer     = nn.Linear(hidden_dim, latent_dim)
        self.logvar_layer = nn.Linear(hidden_dim, latent_dim)
        # Decoder
        self.dec1 = nn.Linear(latent_dim, hidden_dim)
        self.dec2 = nn.Linear(hidden_dim, hidden_dim)
        self.dec_out = nn.Linear(hidden_dim, input_dim)
        self.relu = nn.ReLU()

    def encode(self, x):
        h = self.relu(self.enc1(x))
        h = self.relu(self.enc2(h))
        return self.mu_layer(h), self.logvar_layer(h)

    def reparameterise(self, mu, logvar):
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        return mu + eps * std

    def decode(self, z):
        h = self.relu(self.dec1(z))
        h = self.relu(self.dec2(h))
        return self.dec_out(h)

    def forward(self, x):
        mu, logvar = self.encode(x)
        z = self.reparameterise(mu, logvar)
        recon = self.decode(z)
        return recon, mu, logvar


def vae_loss(recon_x, x, mu, logvar, beta=1.0):
    
    recon_loss = nn.functional.mse_loss(recon_x, x, reduction="sum")
    # KL divergence between learned latent distribution and N(0, I)
    kld = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp())
    return recon_loss + beta * kld, recon_loss, kld


device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"  Using device: {device}")

model = VAE(n_features, HIDDEN_DIM, LATENT_DIM).to(device)
optimizer = optim.Adam(model.parameters(), lr=LEARNING_RATE)
```

```python
# STEP 3 — TRAIN THE VAE
print(f"\n[3/8] Training VAE for {EPOCHS} epochs...")
 
X_tensor = torch.tensor(X)
dataset = TensorDataset(X_tensor)
loader = DataLoader(dataset, batch_size=BATCH_SIZE, shuffle=True)
 
history = {"total": [], "recon": [], "kld": []}
 
model.train()
for epoch in range(1, EPOCHS + 1):
    epoch_total, epoch_recon, epoch_kld = 0.0, 0.0, 0.0
    for (batch,) in loader:
        batch = batch.to(device)
        optimizer.zero_grad()
        recon, mu, logvar = model(batch)
        loss, recon_loss, kld = vae_loss(recon, batch, mu, logvar, beta=BETA)
        loss.backward()
        optimizer.step()
 
        epoch_total += loss.item()
        epoch_recon += recon_loss.item()
        epoch_kld += kld.item()
 
    n = len(dataset)
    history["total"].append(epoch_total / n)
    history["recon"].append(epoch_recon / n)
    history["kld"].append(epoch_kld / n)
 
    if epoch % 50 == 0 or epoch == 1:
        print(f"  Epoch {epoch:4d}/{EPOCHS} | "
              f"total={history['total'][-1]:.3f} | "
              f"recon={history['recon'][-1]:.3f} | "
              f"kld={history['kld'][-1]:.3f}")
 
# Plot training curves
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
axes[0].plot(history["total"], color=PALETTE[0]); axes[0].set_title("Total loss")
axes[1].plot(history["recon"], color=PALETTE[1]); axes[1].set_title("Reconstruction loss")
axes[2].plot(history["kld"], color=PALETTE[2]);   axes[2].set_title("KL divergence")
for ax in axes:
    ax.set_xlabel("Epoch")
    ax.set_ylabel("Loss (per sample)")
plt.tight_layout()
plt.savefig(f"{OUTPUT_DIR}/01_training_curves.png", bbox_inches="tight")
plt.show()
plt.close()
print("  Saved: 01_training_curves.png")
 
```

```python
# STEP 4: EXTRACT LATENT REPRESENTATIONS
print("\n[4/8] Extracting latent vectors (mu) for every child...")

model.eval()
with torch.no_grad():
    mu_all, logvar_all = model.encode(X_tensor.to(device))
    Z = mu_all.cpu().numpy()   # use the mean of the latent distribution (deterministic)

print(f"  Latent matrix shape: {Z.shape} (children x latent dims)")

Z_df = pd.DataFrame(Z, columns=[f"z{i+1}" for i in range(LATENT_DIM)])


Z_df.insert(0, "ID", child_ids.values)
Z_df.to_csv(f"{OUTPUT_DIR}/latent_vectors.csv", index=False)
print(f"  Saved: latent_vectors.csv (includes ID for safe merging)")
```

```python
# STEP 5: UMAP PROJECTION (for visualisation only, clustering is on Z)
print("\n[5/8] Projecting latent space to 2D with UMAP...")

reducer = umap.UMAP(n_neighbors=15, min_dist=0.1, n_components=2,
                     random_state=RANDOM_SEED)
umap_coords = reducer.fit_transform(Z)

umap_df = pd.DataFrame(umap_coords, columns=["umap1", "umap2"])
umap_df.insert(0, "ID", child_ids.values)   # FIX: same ID consistency as Step 4
umap_df.to_csv(f"{OUTPUT_DIR}/umap_coords.csv", index=False)
print("  Saved: umap_coords.csv (includes ID for traceability)")
print(f"  UMAP coordinates shape: {umap_df.shape} (should be 1281 rows)")
```

```python
# STEP 6 — GMM CLUSTERING: SELECT k VIA BIC
print("\n[6/8] Fitting GMM for k=2..8, selecting best k via BIC...")
 
k_range = range(2, 9)
bic_scores, aic_scores, sil_scores, gmm_models = [], [], [], []
 
for k in k_range:
    gmm = GaussianMixture(n_components=k, covariance_type="full",
                           random_state=RANDOM_SEED, n_init=5)
    labels = gmm.fit_predict(Z)
    bic_scores.append(gmm.bic(Z))
    aic_scores.append(gmm.aic(Z))
    sil_scores.append(silhouette_score(Z, labels))
    gmm_models.append(gmm)
    print(f"  k={k}: BIC={bic_scores[-1]:.1f} | AIC={aic_scores[-1]:.1f} | "
          f"silhouette={sil_scores[-1]:.3f}")
 
best_k_idx = int(np.argmin(bic_scores))
best_k = list(k_range)[best_k_idx]
best_gmm = gmm_models[best_k_idx]
cluster_labels = best_gmm.predict(Z)
 
print(f"\n  --> Selected k={best_k} (lowest BIC)")
 
# Plot BIC/AIC/silhouette curves
fig, axes = plt.subplots(1, 2, figsize=(13, 5))
axes[0].plot(list(k_range), bic_scores, "o-", label="BIC", color=PALETTE[0])
axes[0].plot(list(k_range), aic_scores, "o-", label="AIC", color=PALETTE[1])
axes[0].axvline(best_k, linestyle="--", color="grey", label=f"selected k={best_k}")
axes[0].set_xlabel("Number of clusters (k)")
axes[0].set_ylabel("Information criterion")
axes[0].set_title("Model selection: BIC / AIC")
axes[0].legend()
 
axes[1].plot(list(k_range), sil_scores, "o-", color=PALETTE[2])
axes[1].axvline(best_k, linestyle="--", color="grey")
axes[1].set_xlabel("Number of clusters (k)")
axes[1].set_ylabel("Silhouette score")
axes[1].set_title("Cluster separation: silhouette score")
 
plt.tight_layout()
plt.savefig(f"{OUTPUT_DIR}/02_model_selection.png", bbox_inches="tight")
plt.show()
plt.close()
print("  Saved: 02_model_selection.png")
 
```

```python
# STEP 6b — GMM CLUSTERING WITH DIAGONAL COVARIANCE (comparison)
print("\n[6b/8] Re-fitting GMM with covariance_type='diag' (fewer parameters)...")

k_range = range(2, 9)
bic_scores_diag, aic_scores_diag, sil_scores_diag, gmm_models_diag = [], [], [], []

for k in k_range:
    gmm = GaussianMixture(n_components=k, covariance_type="diag",
                           random_state=RANDOM_SEED, n_init=5)
    labels = gmm.fit_predict(Z)
    bic_scores_diag.append(gmm.bic(Z))
    aic_scores_diag.append(gmm.aic(Z))
    sil_scores_diag.append(silhouette_score(Z, labels))
    gmm_models_diag.append(gmm)
    print(f"  k={k}: BIC={bic_scores_diag[-1]:.1f} | AIC={aic_scores_diag[-1]:.1f} | "
          f"silhouette={sil_scores_diag[-1]:.3f}")

best_k_idx_diag = int(np.argmin(bic_scores_diag))
best_k_diag = list(k_range)[best_k_idx_diag]
print(f"\n  --> Selected k={best_k_diag} (lowest BIC, diag covariance)")
print(f"  Compare: full covariance selected k=8 with weak silhouette (~0.03-0.09)")
```

```python
# STEP 6c — BOOTSTRAP STABILITY CHECK: k=2 vs k=3 (diag covariance)
print("\n[6c/8] Bootstrap stability check for k=2 and k=3 (diag covariance)...")

from sklearn.metrics import adjusted_rand_score

def bootstrap_stability(Z, k, n_boot=50, seed=RANDOM_SEED):
    gmm_full = GaussianMixture(n_components=k, covariance_type="diag",
                                random_state=seed, n_init=5)
    labels_full = gmm_full.fit_predict(Z)

    rng = np.random.default_rng(seed)
    n_samples = Z.shape[0]
    ari_scores = []

    for b in range(n_boot):
        idx = rng.choice(n_samples, size=n_samples, replace=True)
        gmm_b = GaussianMixture(n_components=k, covariance_type="diag",
                                 random_state=b, n_init=1)
        labels_b = gmm_b.fit_predict(Z[idx])
        ari = adjusted_rand_score(labels_full[idx], labels_b)
        ari_scores.append(ari)

    return labels_full, np.mean(ari_scores), np.std(ari_scores)

results_stability = {}
for k in [2, 3]:
    labels, ari_mean, ari_std = bootstrap_stability(Z, k)
    sil = silhouette_score(Z, labels)
    results_stability[k] = {"labels": labels, "ari_mean": ari_mean,
                             "ari_std": ari_std, "silhouette": sil}
    print(f"  k={k}: silhouette={sil:.3f} | bootstrap ARI mean={ari_mean:.3f}, "
          f"std={ari_std:.3f}")
    print(f"    Cluster sizes: {np.bincount(labels)}")

print("\n  (ARI > 0.5 suggests reasonably stable clusters; closer to 1.0 is more stable)")

# Recommend based on stability + silhouette combined
best_k_final = max([2, 3], key=lambda k: results_stability[k]["ari_mean"])
print(f"\n  --> More stable option: k={best_k_final} "
      f"(ARI={results_stability[best_k_final]['ari_mean']:.3f})")
```

```python
# ============================================================
# INFANT VAE PIPELINE — CONFIG
# ============================================================
INFANT_INPUT_PATH = "preprocessing_outputs/X_scaled_infant_features.csv"
INFANT_CLEAN_PATH = "preprocessing_outputs/df_clean.csv"
INFANT_OUTPUT_DIR = "vae_infant_outputs"
os.makedirs(INFANT_OUTPUT_DIR, exist_ok=True)

# Smaller architecture — 17 features vs 70, fewer children (606 vs 1,262)
INFANT_LATENT_DIM  = 4
INFANT_HIDDEN_DIM  = 16
INFANT_BETA        = 1.0
INFANT_EPOCHS      = 500
INFANT_BATCH_SIZE  = 32
INFANT_LR          = 1e-3

torch.manual_seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)

print("=" * 70)
print("INFANT VAE + UMAP + GMM CLUSTERING PIPELINE")
print("=" * 70)
```

```python
# STEP 1 — LOAD INFANT DATA
print("\n[1/8] Loading scaled infant diet matrix...")

X_infant_df = pd.read_csv(INFANT_INPUT_PATH)
infant_child_ids = X_infant_df["ID"].copy()
infant_feature_names = [c for c in X_infant_df.columns if c != "ID"]
X_infant = X_infant_df[infant_feature_names].values.astype(np.float32)

print(f"  Loaded: {X_infant.shape[0]:,} infants x {X_infant.shape[1]} diet features")
print(f"  Features: {infant_feature_names}")
print(f"  Missing values: {np.isnan(X_infant).sum()} (should be 0)")

n_features_infant = X_infant.shape[1]
```

```python
# STEP 2 — DEFINE THE INFANT VAE
print("\n[2/8] Building infant VAE architecture...")

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"  Using device: {device}")

model_infant = VAE(n_features_infant, INFANT_HIDDEN_DIM, INFANT_LATENT_DIM).to(device)
optimizer_infant = optim.Adam(model_infant.parameters(), lr=INFANT_LR)
```

```python
# STEP 3 — TRAIN THE INFANT VAE
print(f"\n[3/8] Training infant VAE for {INFANT_EPOCHS} epochs...")

X_tensor_infant = torch.tensor(X_infant)
dataset_infant = TensorDataset(X_tensor_infant)
loader_infant = DataLoader(dataset_infant, batch_size=INFANT_BATCH_SIZE, shuffle=True)

history_infant = {"total": [], "recon": [], "kld": []}

model_infant.train()
for epoch in range(1, INFANT_EPOCHS + 1):
    epoch_total, epoch_recon, epoch_kld = 0.0, 0.0, 0.0
    for (batch,) in loader_infant:
        batch = batch.to(device)
        optimizer_infant.zero_grad()
        recon, mu, logvar = model_infant(batch)
        loss, recon_loss, kld = vae_loss(recon, batch, mu, logvar, beta=INFANT_BETA)
        loss.backward()
        optimizer_infant.step()
        epoch_total += loss.item()
        epoch_recon += recon_loss.item()
        epoch_kld += kld.item()

    n = len(dataset_infant)
    history_infant["total"].append(epoch_total / n)
    history_infant["recon"].append(epoch_recon / n)
    history_infant["kld"].append(epoch_kld / n)

    if epoch % 50 == 0 or epoch == 1:
        print(f"  Epoch {epoch:4d}/{INFANT_EPOCHS} | "
              f"total={history_infant['total'][-1]:.3f} | "
              f"recon={history_infant['recon'][-1]:.3f} | "
              f"kld={history_infant['kld'][-1]:.3f}")

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
axes[0].plot(history_infant["total"], color=PALETTE[0]); axes[0].set_title("Total loss")
axes[1].plot(history_infant["recon"], color=PALETTE[1]); axes[1].set_title("Reconstruction loss")
axes[2].plot(history_infant["kld"], color=PALETTE[2]);   axes[2].set_title("KL divergence")
for ax in axes:
    ax.set_xlabel("Epoch"); ax.set_ylabel("Loss (per sample)")
plt.tight_layout()
plt.savefig(f"{INFANT_OUTPUT_DIR}/01_training_curves.png", bbox_inches="tight")
plt.show()
plt.close()
print("  Saved: 01_training_curves.png")
```

```python
# STEP 4 — EXTRACT INFANT LATENT REPRESENTATIONS
print("\n[4/8] Extracting latent vectors (mu) for every infant...")

model_infant.eval()
with torch.no_grad():
    mu_all_infant, _ = model_infant.encode(X_tensor_infant.to(device))
    Z_infant = mu_all_infant.cpu().numpy()

print(f"  Latent matrix shape: {Z_infant.shape} (infants x latent dims)")

Z_infant_df = pd.DataFrame(Z_infant, columns=[f"z{i+1}" for i in range(INFANT_LATENT_DIM)])
Z_infant_df.insert(0, "ID", infant_child_ids.values)
Z_infant_df.to_csv(f"{INFANT_OUTPUT_DIR}/latent_vectors_infant.csv", index=False)
print(f"  Saved: latent_vectors_infant.csv (includes ID for safe merging)")
```

```python
# STEP 5 — UMAP PROJECTION (infant, visualisation only)
print("\n[5/8] Projecting infant latent space to 2D with UMAP...")

reducer_infant = umap.UMAP(n_neighbors=15, min_dist=0.1, n_components=2, random_state=RANDOM_SEED)
umap_coords_infant = reducer_infant.fit_transform(Z_infant)

umap_infant_df = pd.DataFrame(umap_coords_infant, columns=["umap1", "umap2"])
umap_infant_df.insert(0, "ID", infant_child_ids.values)
umap_infant_df.to_csv(f"{INFANT_OUTPUT_DIR}/umap_coords_infant.csv", index=False)
print(f"  Saved: umap_coords_infant.csv")
print(f"  UMAP coordinates shape: {umap_infant_df.shape} (should be {len(Z_infant)} rows)")
```

```python
# STEP 6 — GMM CLUSTERING (infant): SELECT k VIA BIC, full covariance
print("\n[6/8] Fitting GMM for k=2..6, selecting best k via BIC (full covariance)...")

k_range_infant = range(2, 7)   # smaller range given n=606
bic_i, aic_i, sil_i, models_i = [], [], [], []

for k in k_range_infant:
    gmm = GaussianMixture(n_components=k, covariance_type="full", random_state=RANDOM_SEED, n_init=5)
    labels = gmm.fit_predict(Z_infant)
    bic_i.append(gmm.bic(Z_infant))
    aic_i.append(gmm.aic(Z_infant))
    sil_i.append(silhouette_score(Z_infant, labels))
    models_i.append(gmm)
    print(f"  k={k}: BIC={bic_i[-1]:.1f} | AIC={aic_i[-1]:.1f} | silhouette={sil_i[-1]:.3f}")

print(f"\n  --> full covariance best k (lowest BIC): {list(k_range_infant)[int(np.argmin(bic_i))]}")
```

```python
# STEP 6b — GMM CLUSTERING (infant): diag covariance comparison
print("\n[6b/8] Re-fitting infant GMM with covariance_type='diag'...")

bic_i_diag, aic_i_diag, sil_i_diag, models_i_diag = [], [], [], []

for k in k_range_infant:
    gmm = GaussianMixture(n_components=k, covariance_type="diag", random_state=RANDOM_SEED, n_init=5)
    labels = gmm.fit_predict(Z_infant)
    bic_i_diag.append(gmm.bic(Z_infant))
    aic_i_diag.append(gmm.aic(Z_infant))
    sil_i_diag.append(silhouette_score(Z_infant, labels))
    models_i_diag.append(gmm)
    print(f"  k={k}: BIC={bic_i_diag[-1]:.1f} | AIC={aic_i_diag[-1]:.1f} | silhouette={sil_i_diag[-1]:.3f}")

print(f"\n  --> diag covariance best k (lowest BIC): {list(k_range_infant)[int(np.argmin(bic_i_diag))]}")
```

```python
# STEP 6c — BOOTSTRAP STABILITY CHECK (infant): k=2 vs k=3, diag covariance
print("\n[6c/8] Bootstrap stability check for infant k=2 and k=3 (diag covariance)...")

def bootstrap_stability_infant(Z, k, n_boot=50, seed=RANDOM_SEED):
    gmm_full = GaussianMixture(n_components=k, covariance_type="diag", random_state=seed, n_init=5)
    labels_full = gmm_full.fit_predict(Z)
    rng = np.random.default_rng(seed)
    n_samples = Z.shape[0]
    ari_scores = []
    for b in range(n_boot):
        idx = rng.choice(n_samples, size=n_samples, replace=True)
        gmm_b = GaussianMixture(n_components=k, covariance_type="diag", random_state=b, n_init=1)
        labels_b = gmm_b.fit_predict(Z[idx])
        ari_scores.append(adjusted_rand_score(labels_full[idx], labels_b))
    return labels_full, np.mean(ari_scores), np.std(ari_scores)

results_stability_infant = {}
for k in [2, 3]:
    labels, ari_mean, ari_std = bootstrap_stability_infant(Z_infant, k)
    sil = silhouette_score(Z_infant, labels)
    results_stability_infant[k] = {"labels": labels, "ari_mean": ari_mean, "ari_std": ari_std, "silhouette": sil}
    print(f"  k={k}: silhouette={sil:.3f} | bootstrap ARI mean={ari_mean:.3f}, std={ari_std:.3f}")
    print(f"    Cluster sizes: {np.bincount(labels)}")

print("\n  (ARI > 0.5 suggests reasonably stable clusters)")
```

```python
# STEP 1 — LOAD DATA
print("\n[1/6] Loading clustered dataset...")
df = pd.read_csv(INPUT_PATH, low_memory=False)
print(f"  Loaded: {df.shape[0]:,} children x {df.shape[1]} columns")

n_older = df[CLUSTER_COL].notna().sum()
n_infant = df[INFANT_CLUSTER_COL].notna().sum()
n_neither = ((df[CLUSTER_COL].isna()) & (df[INFANT_CLUSTER_COL].isna())).sum()
print(f"  Children with dietary_cluster (older, 2-13y): {n_older:,}")
print(f"  Children with infant_dietary_cluster (6-24m): {n_infant:,}")
print(f"  Children with NEITHER (unexpected — check pipeline): {n_neither:,}")

# Older-group cluster: labelled category, cluster 0 = reference group
n_clusters_older = df[CLUSTER_COL].nunique()
print(f"  Dietary clusters (older) found: {n_clusters_older} -> "
      f"{sorted(df.loc[df[CLUSTER_COL].notna(), CLUSTER_COL].unique())}")
df["cluster_cat"] = pd.Categorical(
    df[CLUSTER_COL],
    categories=sorted(df.loc[df[CLUSTER_COL].notna(), CLUSTER_COL].astype(int).unique())
)

# Infant-group cluster: same treatment, separate categorical column
n_clusters_infant = df[INFANT_CLUSTER_COL].nunique()
print(f"  Dietary clusters (infant) found: {n_clusters_infant} -> "
      f"{sorted(df.loc[df[INFANT_CLUSTER_COL].notna(), INFANT_CLUSTER_COL].unique())}")
df["infant_cluster_cat"] = pd.Categorical(
    df[INFANT_CLUSTER_COL],
    categories=sorted(df.loc[df[INFANT_CLUSTER_COL].notna(), INFANT_CLUSTER_COL].astype(int).unique())
)
```

```python
# STEP 2 — CLEAN COVARIATES
print("\n[2/6] Preparing covariates...")

for col in ["agey", "income1", "caretakeredu"]:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

df["gender_clean"] = df["gender"].astype(str).str.strip()
df = df[df["gender_clean"].isin(["Male", "Female"])].copy()
df["sex_male"] = (df["gender_clean"] == "Male").astype(int)

def haz_to_stunting(haz):
    if pd.isna(haz):
        return np.nan
    if haz < -3:
        return 2
    elif haz < -2:
        return 1
    else:
        return 0
df["stunting_ord"] = df["haz"].apply(haz_to_stunting)

df[AGEGROUP_COL] = df[AGEGROUP_COL].astype(str).str.strip()
df_asq = df[df[AGEGROUP_COL] == "6-24 months"].copy()
df_older = df[df[AGEGROUP_COL].isin(["2-5 years", "6-13 years"])].copy()

print(f"  ASQ subgroup   : {len(df_asq):,} children")
print(f"  Older subgroup : {len(df_older):,} children")
```

```python
# STEP 3 — MODEL FITTING FUNCTION
print("\n[3/6] Fitting models for each cognitive outcome...")

results = []

def fit_ols_model(data, outcome_col, label, cluster_col_name):
    """OLS sensitivity model: outcome ~ cluster + sex + age + income + stunting"""
    d = data[[outcome_col, cluster_col_name, "sex_male", "agey",
              "income1", "stunting_ord"]].dropna()
    if len(d) < 30 or d[cluster_col_name].nunique() < 2:
        print(f"  SKIP (insufficient data): {label}")
        return None, d
    d = d.copy()
    d["cluster_cat_model"] = pd.Categorical(
        d[cluster_col_name].astype(int),
        categories=sorted(d[cluster_col_name].dropna().astype(int).unique())
    )
    formula = (f"{outcome_col} ~ C(cluster_cat_model) + sex_male + agey + "
               f"income1 + stunting_ord")
    model = smf.ols(formula, data=d).fit()
    return model, d


def fit_gam_model(data, outcome_col, cluster_col_name):
    """GAM sensitivity model: same predictors as OLS, but age modelled
    flexibly via a spline instead of assuming a linear effect."""
    cols = [outcome_col, cluster_col_name, "sex_male", "agey", "income1", "stunting_ord"]
    d = data[cols].dropna()
    if len(d) < 30 or d[cluster_col_name].nunique() < 2:
        return None, d
    d = d.copy()
    d[cluster_col_name] = d[cluster_col_name].astype(int)
    d["stunting_ord"] = d["stunting_ord"].astype(int)

    X = d[[cluster_col_name, "sex_male", "agey", "income1", "stunting_ord"]].values
    y = d[outcome_col].values

    # f() = factor term (cluster, sex, stunting — few discrete levels)
    # s() = spline term (agey, income1 — allow non-linear effects)
    terms = f(0) + f(1) + s(2) + s(3) + f(4)
    try:
        gam = LinearGAM(terms).fit(X, y)
        return gam, d
    except Exception as e:
        print(f"     GAM fit failed: {e}")
        return None, d


def process_outcome_group(data, outcome_dict, subgroup_label, cluster_col_name):
    for label, col in outcome_dict.items():
        if col not in data.columns:
            print(f"  SKIP (column not found): {label} ({col})")
            continue

        print(f"\n  -- {subgroup_label} | {label} ({col}) --")
        ols_model, d = fit_ols_model(data, col, label, cluster_col_name)
        if ols_model is None:
            continue

        print(f"     n={len(d)}, R^2={ols_model.rsquared:.3f}, "
              f"adj R^2={ols_model.rsquared_adj:.3f}")

        for pname, coef, pval, ci in zip(
            ols_model.params.index, ols_model.params.values,
            ols_model.pvalues.values, ols_model.conf_int().values
        ):
            if "cluster_cat_model" in pname:
                cluster_id = pname.split("T.")[-1].strip("]")
                results.append({
                    "subgroup": subgroup_label,
                    "outcome": label,
                    "outcome_col": col,
                    "cluster_vs_ref": cluster_id,
                    "coef": coef,
                    "p_value": pval,
                    "ci_low": ci[0],
                    "ci_high": ci[1],
                    "n": len(d),
                    "r_squared": ols_model.rsquared,
                })
                sig = "***" if pval < 0.01 else ("**" if pval < 0.05 else
                      ("*" if pval < 0.1 else ""))
                print(f"       Cluster {cluster_id} vs ref: "
                      f"coef={coef:+.3f}, p={pval:.3f} {sig}")

        # pass cluster_col_name through — was missing an argument
        if PYGAM_AVAILABLE:
            gam_model, _ = fit_gam_model(data, col, cluster_col_name)
            if gam_model is not None:
                print(f"     GAM pseudo-R^2: {gam_model.statistics_['pseudo_r2']['explained_deviance']:.3f}")


process_outcome_group(df_asq, ASQ_OUTCOMES, "ASQ (6-24 months)", "infant_cluster_cat")
process_outcome_group(df_older, OLDER_OUTCOMES, "Older (2-13 years)", "cluster_cat")

results_df = pd.DataFrame(results)
results_df.to_csv(f"{OUTPUT_DIR}/cluster_regression_results.csv", index=False)
print(f"\n  Saved: cluster_regression_results.csv ({len(results_df)} rows)")
```

```python
# STEP 3b — MULTIPLE COMPARISON CORRECTION
print("\n[3b/6] Applying Benjamini-Hochberg correction across all 9 tests...")
reject, pvals_corrected, _, _ = multipletests(results_df["p_value"], method="fdr_bh")
results_df["p_value_corrected"] = pvals_corrected
results_df["significant_corrected"] = reject
print(results_df[["subgroup", "outcome", "p_value", "p_value_corrected", "significant_corrected"]]
      .to_string(index=False))
results_df.to_csv(f"{OUTPUT_DIR}/cluster_regression_results.csv", index=False)
```

```python
# STEP 7 — MANOVA: JOINT TEST ACROSS COGNITIVE DOMAINS
print("\n[7/8] MANOVA — joint test of cluster effect across cognitive domains...")

covariates_model = ["sex_male", "agey", "income1", "stunting_ord"]

def fit_manova(data, outcome_cols, cluster_col, covariates, label):
    cols = outcome_cols + [cluster_col] + covariates
    cols = [c for c in cols if c in data.columns]
    d = data[cols].dropna().copy()

    if len(d) < 30 or d[cluster_col].nunique() < 2:
        print(f"  SKIP (insufficient data): {label} (n={len(d)})")
        return None, d

    d[cluster_col] = d[cluster_col].astype(int)
    lhs = " + ".join(outcome_cols)
    rhs = f"C({cluster_col}) + " + " + ".join(covariates)
    formula = f"{lhs} ~ {rhs}"

    print(f"\n  -- {label} (n={len(d)}) --")
    print(f"     Formula: {lhs} ~ {rhs}")

    manova = MANOVA.from_formula(formula, data=d)
    mv_result = manova.mv_test()
    return mv_result, d

def print_manova_cluster_effect(mv_result, cluster_col):
    """Extract and print just the cluster term's test statistics."""
    term_name = f"C({cluster_col})"
    try:
        table = mv_result.results[term_name]["stat"]
        print(table.to_string())
        wilks_p = table.loc["Wilks' lambda", "Pr > F"]
        sig = "***" if wilks_p < 0.01 else ("**" if wilks_p < 0.05 else ("*" if wilks_p < 0.10 else ""))
        print(f"\n     Wilks' lambda p-value for cluster effect: {wilks_p:.4f} {sig}")
        return wilks_p
    except KeyError:
        print(f"     Could not find term '{term_name}' in MANOVA results — check output above.")
        return None

manova_results = []

# ASQ (infant) — joint test across all 5 ASQ domains
mv_asq, d_asq = fit_manova(
    df_asq, list(ASQ_OUTCOMES.values()), "infant_cluster_cat",
    covariates_model, "ASQ (6-24 months) — all 5 domains jointly"
)
if mv_asq is not None:
    p_asq = print_manova_cluster_effect(mv_asq, "infant_cluster_cat")
    manova_results.append({"subgroup": "ASQ (6-24 months)", "n": len(d_asq),
                            "n_outcomes": len(ASQ_OUTCOMES), "wilks_p": p_asq})

# Older — joint test across all 4 older cognitive tests
mv_older, d_older = fit_manova(
    df_older, list(OLDER_OUTCOMES.values()), "cluster_cat",
    covariates_model, "Older (2-13 years) — all 4 cognitive tests jointly"
)
if mv_older is not None:
    p_older = print_manova_cluster_effect(mv_older, "cluster_cat")
    manova_results.append({"subgroup": "Older (2-13 years)", "n": len(d_older),
                            "n_outcomes": len(OLDER_OUTCOMES), "wilks_p": p_older})

manova_df = pd.DataFrame(manova_results)
manova_df.to_csv(f"{OUTPUT_DIR}/manova_results.csv", index=False)
print(f"\n  Saved: manova_results.csv")
```

```python
# STEP 8 — MALNUTRITION × DIETARY CLUSTER INTERACTION (RQ3)
print("\n[8/8] Testing whether malnutrition (stunting) modifies the "
      "diet-cluster -> cognition relationship...")

interaction_results = []

def fit_interaction_model(data, outcome_col, cluster_col, label):
    """Nested model comparison: does adding cluster x stunting interaction
    significantly improve fit over the main-effects-only model?"""
    cols = [outcome_col, cluster_col, "sex_male", "agey", "income1", "stunting_ord"]
    d = data[cols].dropna().copy()

    if len(d) < 30 or d[cluster_col].nunique() < 2 or d["stunting_ord"].nunique() < 2:
        print(f"  SKIP (insufficient data or variation): {label} (n={len(d)})")
        return None

    d[cluster_col] = pd.Categorical(d[cluster_col].astype(int))
    d["stunting_ord"] = pd.Categorical(d["stunting_ord"].astype(int))

    # Check every cluster x stunting cell has at least 1 observation
    ct = pd.crosstab(d[cluster_col], d["stunting_ord"])
    if (ct == 0).any().any():
        print(f"  SKIP (empty cluster x stunting_ord cell): {label} (n={len(d)})")
        print(ct)
        return None

    # Model A: main effects only
    formula_main = f"{outcome_col} ~ C({cluster_col}) + sex_male + agey + income1 + C(stunting_ord)"
    # Model B: main effects + interaction
    formula_int = f"{outcome_col} ~ C({cluster_col}) * C(stunting_ord) + sex_male + agey + income1"

    model_main = smf.ols(formula_main, data=d).fit()
    model_int = smf.ols(formula_int, data=d).fit()

    # Nested F-test: does the interaction term significantly improve fit?
    anova_result = sm.stats.anova_lm(model_main, model_int)
    f_stat = anova_result["F"].iloc[1]
    p_val = anova_result["Pr(>F)"].iloc[1]

    print(f"\n  -- {label} (n={len(d)}) --")
    print(f"     Main effects R^2={model_main.rsquared:.3f}  |  "
          f"With interaction R^2={model_int.rsquared:.3f}")
    print(f"     Interaction F-test: F={f_stat:.3f}, p={p_val:.4f}")

    return {"outcome": label, "n": len(d),
            "r2_main": model_main.rsquared, "r2_interaction": model_int.rsquared,
            "f_stat": f_stat, "p_value": p_val}

# ASQ (infant) interaction tests
for label, col in ASQ_OUTCOMES.items():
    if col in df_asq.columns:
        res = fit_interaction_model(df_asq, col, "infant_cluster_cat", f"ASQ | {label}")
        if res:
            res["subgroup"] = "ASQ (6-24 months)"
            interaction_results.append(res)

# Older interaction tests
for label, col in OLDER_OUTCOMES.items():
    if col in df_older.columns:
        res = fit_interaction_model(df_older, col, "cluster_cat", f"Older | {label}")
        if res:
            res["subgroup"] = "Older (2-13 years)"
            interaction_results.append(res)

interaction_df = pd.DataFrame(interaction_results)

# apply multiple-comparison correction only to valid (non-NaN) p-values, so one failed model doesn't silently corrupt every other test's correction
valid_mask = interaction_df["p_value"].notna()
interaction_df["p_value_corrected"] = np.nan
interaction_df["significant_corrected"] = False
if valid_mask.sum() > 0:
    reject, pvals_corrected, _, _ = multipletests(
        interaction_df.loc[valid_mask, "p_value"], method="fdr_bh"
    )
    interaction_df.loc[valid_mask, "p_value_corrected"] = pvals_corrected
    interaction_df.loc[valid_mask, "significant_corrected"] = reject
n_skipped = (~valid_mask).sum()
if n_skipped > 0:
    print(f"  NOTE: {n_skipped} test(s) could not be fit (likely an empty "
          f"cluster x stunting cell) — excluded from correction, not counted "
          f"as null results")

interaction_df.to_csv(f"{OUTPUT_DIR}/malnutrition_interaction_results.csv", index=False)
print("\n  Summary — malnutrition (stunting) x dietary cluster interaction effects:")
print(interaction_df[["subgroup", "outcome", "n", "f_stat", "p_value", "p_value_corrected"]]
      .to_string(index=False))
```

```python
# STEP 8b — INTERACTION TEST USING THINNESS (OLDER) / WASTING (INFANT)
print("\n[8b/8] Re-testing malnutrition x cluster interaction using "
      "thinness3 (older) and wasting3 (infants) — more prevalent than stunting...")

for col in ["thinness3_num", "wasting3_num"]:
    if col not in df_older.columns and col not in df_asq.columns:
        print(f"  WARNING: {col} not found — check it survived the merge into "
              f"df_clean_with_clusters.csv")

def fit_interaction_model_v2(data, outcome_col, cluster_col, malnut_col, label):
    """Same nested F-test as before, but with a configurable malnutrition
    indicator instead of hardcoding stunting_ord."""
    cols = [outcome_col, cluster_col, "sex_male", "agey", "income1", malnut_col]
    d = data[cols].dropna().copy()

    if len(d) < 30 or d[cluster_col].nunique() < 2 or d[malnut_col].nunique() < 2:
        print(f"  SKIP (insufficient data or variation): {label} (n={len(d)})")
        return None

    d[cluster_col] = pd.Categorical(d[cluster_col].astype(int))
    d[malnut_col] = pd.Categorical(d[malnut_col].astype(int))

    # Check every cluster x malnutrition cell has at least 1 observation
    ct = pd.crosstab(d[cluster_col], d[malnut_col])
    if (ct == 0).any().any():
        print(f"  SKIP (empty cluster x {malnut_col} cell): {label} (n={len(d)})")
        print(ct)
        return None

    formula_main = f"{outcome_col} ~ C({cluster_col}) + sex_male + agey + income1 + C({malnut_col})"
    formula_int  = f"{outcome_col} ~ C({cluster_col}) * C({malnut_col}) + sex_male + agey + income1"

    model_main = smf.ols(formula_main, data=d).fit()
    model_int = smf.ols(formula_int, data=d).fit()
    anova_result = sm.stats.anova_lm(model_main, model_int)
    f_stat = anova_result["F"].iloc[1]
    p_val = anova_result["Pr(>F)"].iloc[1]

    print(f"\n  -- {label} (n={len(d)}) --")
    print(f"     Main R^2={model_main.rsquared:.3f}  |  With interaction R^2={model_int.rsquared:.3f}")
    print(f"     Interaction F-test: F={f_stat:.3f}, p={p_val:.4f}")

    return {"outcome": label, "malnutrition_indicator": malnut_col, "n": len(d),
            "r2_main": model_main.rsquared, "r2_interaction": model_int.rsquared,
            "f_stat": f_stat, "p_value": p_val}

interaction_v2_results = []

# Older group — use thinness3_num (more prevalent than stunting: 20 thin + 301 overweight)
for label, col in OLDER_OUTCOMES.items():
    if col in df_older.columns:
        res = fit_interaction_model_v2(df_older, col, "cluster_cat", "thinness3_num", f"Older | {label}")
        if res:
            res["subgroup"] = "Older (2-13 years)"
            interaction_v2_results.append(res)

# Infants — use wasting3_num (still limited, but better than stunting: 3 wasted + 92 overweight)
for label, col in ASQ_OUTCOMES.items():
    if col in df_asq.columns:
        res = fit_interaction_model_v2(df_asq, col, "infant_cluster_cat", "wasting3_num", f"ASQ | {label}")
        if res:
            res["subgroup"] = "ASQ (6-24 months)"
            interaction_v2_results.append(res)

interaction_v2_df = pd.DataFrame(interaction_v2_results)

# Multiple-comparison correction, NaN-safe
if not interaction_v2_df.empty:
    valid_mask = interaction_v2_df["p_value"].notna()
    interaction_v2_df["p_value_corrected"] = np.nan
    interaction_v2_df["significant_corrected"] = False
    if valid_mask.sum() > 0:
        reject, pvals_corrected, _, _ = multipletests(
            interaction_v2_df.loc[valid_mask, "p_value"], method="fdr_bh"
        )
        interaction_v2_df.loc[valid_mask, "p_value_corrected"] = pvals_corrected
        interaction_v2_df.loc[valid_mask, "significant_corrected"] = reject

interaction_v2_df.to_csv(f"{OUTPUT_DIR}/malnutrition_interaction_v2_results.csv", index=False)
print(f"\n  Saved: malnutrition_interaction_v2_results.csv ({len(interaction_v2_df)} rows)")
print("\n  Summary — thinness/wasting x dietary cluster interaction effects:")
print(interaction_v2_df[["subgroup", "outcome", "malnutrition_indicator", "n",
                          "f_stat", "p_value", "p_value_corrected"]].to_string(index=False))
```

```python
# STEP 9 — SENSITIVITY ANALYSIS: IMPUTED vs COMPLETE-CASE DIET DATA
print("\n[9/9] Sensitivity analysis: does restricting to children with fully "
      "observed (non-imputed) diet data change the results?")

missing_flag_cols = [c for c in df.columns if c.endswith("_was_missing")]
print(f"  Missingness flags found: {missing_flag_cols}")

for subset in [df_asq, df_older]:
    missing_cols_present = [c for c in missing_flag_cols if c in subset.columns]
    if missing_cols_present:
        subset["any_diet_imputed"] = (subset[missing_cols_present].sum(axis=1) > 0).astype(int)
    else:
        subset["any_diet_imputed"] = np.nan
        print(f"  WARNING: no _was_missing columns found in this subset")

print(f"  Older group: any_diet_imputed present -> {df_older['any_diet_imputed'].notna().sum()} rows checked")
print(f"  ASQ group:   any_diet_imputed present -> {df_asq['any_diet_imputed'].notna().sum()} rows checked")

df_older_complete = df_older[df_older["any_diet_imputed"] == 0].copy()
df_asq_complete = df_asq[df_asq["any_diet_imputed"] == 0].copy()

print(f"  Older group: {len(df_older)} imputed-included vs {len(df_older_complete)} complete-case-only")
print(f"  ASQ group:   {len(df_asq)} imputed-included vs {len(df_asq_complete)} complete-case-only")

sensitivity_results = []

def compare_imputed_vs_complete(data_full, data_complete, outcome_col, cluster_col, label):
    model_full, d_full = fit_ols_model(data_full, outcome_col, label, cluster_col)
    if model_full is None:
        return None
    model_complete, d_complete = fit_ols_model(data_complete, outcome_col, label, cluster_col)

    def get_cluster_coef(model):
        for pname, coef, pval in zip(model.params.index, model.params.values, model.pvalues.values):
            if "cluster_cat_model" in pname:
                return coef, pval
        return None, None

    coef_full, p_full = get_cluster_coef(model_full)
    coef_complete, p_complete = get_cluster_coef(model_complete) if model_complete is not None else (None, None)

    print(f"\n  -- {label} --")
    print(f"     Imputed-included  (n={len(d_full)}): coef={coef_full:+.3f}, p={p_full:.3f}")
    if coef_complete is not None:
        print(f"     Complete-case only (n={len(d_complete)}): coef={coef_complete:+.3f}, p={p_complete:.3f}")
        direction_match = "YES" if (coef_full * coef_complete) > 0 else "NO"
        print(f"     Same direction: {direction_match}")
    else:
        print(f"     Complete-case only: SKIPPED (insufficient data, n={len(d_complete)})")

    return {"outcome": label, "n_imputed": len(d_full), "n_complete": len(d_complete),
            "coef_imputed": coef_full, "p_imputed": p_full,
            "coef_complete": coef_complete, "p_complete": p_complete}

for label, col in ASQ_OUTCOMES.items():
    if col in df_asq.columns:
        res = compare_imputed_vs_complete(df_asq, df_asq_complete, col, "infant_cluster_cat", f"ASQ | {label}")
        if res:
            res["subgroup"] = "ASQ (6-24 months)"
            sensitivity_results.append(res)

for label, col in OLDER_OUTCOMES.items():
    if col in df_older.columns:
        res = compare_imputed_vs_complete(df_older, df_older_complete, col, "cluster_cat", f"Older | {label}")
        if res:
            res["subgroup"] = "Older (2-13 years)"
            sensitivity_results.append(res)

sensitivity_df = pd.DataFrame(sensitivity_results)
sensitivity_df.to_csv(f"{OUTPUT_DIR}/sensitivity_imputed_vs_complete.csv", index=False)
print(f"\n  Saved: sensitivity_imputed_vs_complete.csv ({len(sensitivity_df)} rows)")
```
